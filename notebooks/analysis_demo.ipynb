# Strength Form Analyzer - Analysis Demo

This notebook loads the `results.json` produced by `python -m src.main ...` and explores it with pandas/matplotlib. It does not re-run pose estimation -- run the CLI first:

```bash
python -m src.main --video data/input/squat.mp4 --exercise squat --output outputs/demo
```

````python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

RESULTS_PATH = Path("../outputs/demo/results.json")

with RESULTS_PATH.open() as f:
    results = json.load(f)

print(f"Exercise: {results['exercise']}")
print(f"Summary: {results['summary']}")
````

````python
reps = pd.json_normalize(results["reps"])
reps.head()
````

````python
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(reps["rep_number"], reps["form.overall_score"])
ax.axhline(80, color="black", linestyle="--", linewidth=1, label="good threshold (default)")
ax.set_xlabel("Rep number")
ax.set_ylabel("Form score (0-100)")
ax.set_title("Form score by rep")
ax.legend()
plt.show()
````

````python
issue_rows = [
    {"rep": rep["rep_number"], **issue}
    for rep in results["reps"]
    for issue in rep["form"]["issues"]
]
issues_df = pd.DataFrame(issue_rows)
issues_df
````
